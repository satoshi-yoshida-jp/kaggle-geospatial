```python
customers = pd.read_csv("../../data/customers.csv", index_col=0)

customers.head()
```

```python
orders = pd.read_csv("../../data/orders.csv", index_col=0)
orders.head()
```

```python
remean_age = customers.age - customers.age.mean()
```

```python
def stars(row):
    if row.age >= 50:
        return 3
    elif row.age >= 30:
        return 2
    else:
        return 1

star_ratings = customers.apply(stars, axis="columns")
```

```python
orders.groupby("category").category.count()
orders.groupby("category").unit_price.min()
```

```python
orders.groupby("category").unit_price.min()
```

```python
orders.groupby("product").apply(lambda df: df.payment_method.iloc[0])
```

```python
orders.groupby(["customer_id", "product"]).apply(
    lambda df: df.loc[df.unit_price.idxmax()]
)
```
